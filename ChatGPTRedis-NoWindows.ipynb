```python

# Ejemplo de integración de Redis con IA Generativa
# Simulación de una cola de prompts y cacheo de respuestas

import redis
import time

# Conexión a Redis (localhost por defecto)
r = redis.Redis(host='localhost', port=6379, db=0)

# --- PRODUCTOR: agrega un prompt a la cola ---
def enviar_prompt(prompt):
    r.lpush('prompt_queue', prompt)
    print(f"Prompt enviado a la cola: {prompt}")

# --- CONSUMIDOR: simula procesamiento GPT y guarda respuesta ---
def procesar_prompts():
    print("Esperando prompts...")
    while True:
        item = r.brpop('prompt_queue')
        if item:
            _, prompt = item
            prompt = prompt.decode()
            # Simula una respuesta generativa
            respuesta = f"[GPT] Respuesta simulada a: '{prompt}'"
            r.set(prompt, respuesta, ex=300)  # Cachea con TTL
            print(f"Procesado: {respuesta}")
        time.sleep(1)

# --- VERIFICADOR: intenta obtener respuesta desde la caché ---
def obtener_respuesta(prompt):
    cached = r.get(prompt)
    if cached:
        print("Respuesta desde caché:", cached.decode())
    else:
        print("No encontrada en caché. Se necesita procesar el prompt.")

# --- USO EJEMPLO ---
if __name__ == "__main__":
    # Enviar un prompt
    enviar_prompt("Dame una lista de ideas para startups")
    
    # Verificar si hay una respuesta cacheada
    obtener_respuesta("Dame una lista de ideas para startups")
    
    # Procesar en bucle (comenta si no quieres que corra infinitamente)
    # procesar_prompts()
```
